## Now Let The Cleaning Started

```python
df["Product_Name"] = df["Product_Name"].str.title()
df["Category"] = df["Category"].str.title()
df["Supplier"] = df["Supplier"].str.lower()
df
```

```python
df["Supplier"] = df["Supplier"].replace({
    "tech source": "techsource",
    "key world": "keyworld",
    "cable pro": "cablepro"
})
df
```

```python
# df["Product_Name"].isnull().sum()

```

```python
df["Product_Name"] = df["Product_Name"].fillna("Unknown Product")
df
```

```python
def clean_quantity(value):
    if value == "twenty":
        return 20
    elif isinstance(value, (int, float)):
        return value
    elif isinstance(value, str) and value.isdigit():
        return int(value)
    else:
        return np.nan

df['Quantity'] = df["Quantity"].apply(clean_quantity)
df["Quantity"]
```

```python
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")
df["Reorder_Level"] = pd.to_numeric(df["Reorder_Level"], errors="coerce")
# df["Unit_Price"].head()
# df["Reorder_Level"].head()
df.dtypes
```

```python
# Check the negative quantity
df[df["Quantity"] < 0]
```

```python
df.loc[df["Quantity"] < 0, "Quantity"] = np.nan
# df["Quantity"]
```

```python
df.loc[df["Unit_Price"] < 0, "Unit_Price"] = np.nan
# df["Unit_Price"]
```

## Fill the missing Numeric Value

```python
# For Quantity
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Quantity"].head(7)
```

```python
# For Price
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
# df["Unit_Price"]
```

```python
# df["Last_Update"]
```

```python
from dateutil import parser

def clean_date(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT

df["Last_Update"] = df["Last_Update"].apply(clean_date)
df["Last_Update"]
```

```python
median_date = df["Last_Update"].median()

df["Last_Update"] = df["Last_Update"].fillna(median_date)
df["Last_Update"]
```

## Summary Of Insight

The analysis identified significant data quality issues in the original inventoy dataset used which include missing values, duplicate record, inconsistence capitalization negative quantities and prices and improperly formated date as well. These shows that the raw dataset is messy and can't be used for business analysis without cleaning or preprocessing it.
So after cleaning and standardizing the data, 7 products are identified to be below the reorder levels, these product represent immediate inventory replenishment priorites because if sales continues without restocking them it could later results to out of stocks
The supplier analysis identified that techsource as a supplier has the highest number of product records. Even though this might indicate a strong supplier relationship it also creates potential supplier concentration risk, if the business overly dependent on one supplier.
Electronics record the highest average unit price among the product categories. The most expensive product were also electronic products.

Overall, the analysis shows that the company's immmediate operational concern is inventory replenishment, while its broader data-management concern is data quality. The business should ensure validation rules for the product, quantities,prices, date, supplier and other of the product identifiers. Automated reorder alert will actully go a long way and also standardized supplier records will all reduced the likelihood of stock-out and improve the reliability of the future inventory reports
