```python
import numpy as np
```

```python
np.random.seed(42)
consumption = np.random.randint(200,601, size=120)
```

```python
consumption
# print(len(consumption))
```

## Doing the basic statistics

```python
mean = np.mean(consumption)
median = np.median(consumption)
minimum = np.min(consumption)
maximum = np.max(consumption)
std = np.std(consumption)
```

```python
print('Mean:', mean)
print('Median:', median)
print('Minimum:', minimum)
print('Maximum:', maximum)
print('Standard Deviation:', std)
```

## Identify The Extremes

```python
highest_day = np.argmax(consumption) + 1
lowest_day = np.argmin(consumption) + 1

print(highest_day)
print(lowest_day)
```

```python
print("Highest consumption:", maximum, "kWh")
print("Highest consumption Day is Day", highest_day)
print("Lowest consumption:", minimum, "kWh")
print("Lowest consumption Day is Day", lowest_day)
```

## Monthly Splitting

```python
# Calculating the days, 120 days divided by 4 months give us 30 days per month 

month1 = consumption[:30]
month2 = consumption[30:60]
month3 = consumption[60:90]
month4 = consumption[90:120]
# print(month1)
# print(month3)
# print(month4)
```

```python
print("Month 1", month1)
print("Month 2", month2)
print("Month 3", month3)
print("Month 4", month4)
```

## Monthly Analysis

```python
# For month 1
total_month1 = np.sum(month1)
average_month1 = np.mean(month1)

# month 2
total_month2 = np.sum(month2)
average_month2 = np.mean(month2)

#month 3
total_month3 = np.sum(month3)
average_month3 = np.mean(month3)

# month 4
total_month4 = np.sum(month4)
average_month4 = np.mean(month4)

print(total_month1)
```

```python
print("Month 1 Total", total_month1)
print("Month 1 Average", average_month1)
print("Month 2 Total", total_month2)
print("Month 2 Average", average_month2)
print("Month 3 Total", total_month3)
print("Month 3 Average", average_month3)
print("Month 4 Total", total_month4)
print("Month 4 Average", average_month4)
```

## Weekly Analysis

```python
weekly_data = consumption[:119]

weeks = weekly_data.reshape(17,7)
print(weeks)
# print(weekly_data)
```

```python
weekly_total = np.sum(weeks, axis=1)
weekly_average = np.mean(weeks, axis=1)
print("Weekly Total:",weekly_total)
print("Weekly Average:",weekly_average)
```

## Overall consumption Trend

```python
cumulative_consumption = np.cumsum(consumption)
# print(cumulative_consumption)

print(cumulative_consumption[:10])
```

## Anomaly Detection

```python
upper_limit = mean + (2 * std)
lower_limit = mean - (2 * std)

print(upper_limit, lower_limit)
```

```python
anomalies = (consumption > upper_limit) | (consumption < lower_limit)
anomaly_days = np.where(anomalies)[0] + 1
# print(anomalies)

print(anomaly_days)
print("Anomaly values", consumption[anomalies])
```
